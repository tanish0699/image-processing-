```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("sample.webp")

# OpenCV reads image as BGR
B, G, R = cv2.split(img)

# Mean of R, G and B
gray = (R.astype(float) + G.astype(float) + B.astype(float)) / 3

gray = gray.astype("uint8")

cv2_imshow(img)
cv2_imshow(gray)
```
