```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(6, 5))
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Log Transformation
c = 1

log_img = c * np.log(1 + img.astype(np.float64))

# Convert result to 0-255
log_img = cv2.normalize(log_img, None, 0, 255, cv2.NORM_MINMAX)
log_img = log_img.astype(np.uint8)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_img, cmap='gray')
plt.title("Log Transformed Image")
plt.axis("off")

plt.show()
```

```python
# Power Law / Gamma Transformation

gamma = float(input("Enter the value of gamma: "))
c = 1

# Normalize pixel values to 0-1
normalized_img = img.astype(np.float64) / 255.0

# Apply gamma transformation
gamma_img = c * np.power(normalized_img, gamma)

# Convert back to 0-255
gamma_img = np.clip(gamma_img * 255, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gamma_img, cmap='gray')
plt.title("Gamma Transformed Image")
plt.axis("off")

plt.show()
```

```python
# Contrast Stretching

r_min = int(input("Enter input minimum intensity: "))
r_max = int(input("Enter input maximum intensity: "))

s_min = int(input("Enter output minimum intensity: "))
s_max = int(input("Enter output maximum intensity: "))

# Convert image to float
stretch_img = img.astype(np.float64)

# Apply contrast stretching
stretch_img = ((stretch_img - r_min) * (s_max - s_min)) / (r_max - r_min) + s_min

# Keep pixel values within 0-255
stretch_img = np.clip(stretch_img, 0, 255)

# Convert to uint8
stretch_img = stretch_img.astype(np.uint8)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stretch_img, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.show()
```
