```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("sample.webp", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Image not found!")
else:
    print("Image loaded successfully")
    print("Image size:", img.shape)
```

```python
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off")

plt.show()
```

```python
width = int(input("Enter border width in pixels: "))
color = int(input("Enter border color (0-255): "))

gray_border = cv2.copyMakeBorder(
    img,
    width, width, width, width,
    cv2.BORDER_CONSTANT,
    value=color
)

plt.imshow(gray_border, cmap='gray')
plt.title("Grayscale Image with Border")
plt.axis("off")
plt.show()
```

```python
width = int(input("Enter border width in pixels: "))
color = int(input("Enter border color (0 for black / 255 for white): "))

binary_border = cv2.copyMakeBorder(
    binary,
    width, width, width, width,
    cv2.BORDER_CONSTANT,
    value=color
)

plt.imshow(binary_border, cmap='gray')
plt.title("Binary Image with Border")
plt.axis("off")
plt.show()
```

```python
gray_complement = 255 - img

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_complement, cmap='gray')
plt.title("Grayscale Complement")
plt.axis("off")

plt.show()
```

```python
binary_complement = 255 - binary

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(binary, cmap='gray')
plt.title("Original Binary Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary_complement, cmap='gray')
plt.title("Binary Complement")
plt.axis("off")

plt.show()
```
